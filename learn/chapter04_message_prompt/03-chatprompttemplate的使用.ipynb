# 1.提示词模板的三种调用方式

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage
from dotenv import load_dotenv
import os

load_dotenv(override=True)

# 初始化模型
model = init_chat_model(
    model="glm-4.5-air",
    model_provider="openai",
    api_key=os.getenv("ZHIPUAI_API_KEY"),
    base_url=os.getenv("ZHIPUAI_BASE_URL")
)

# 提示词模板
chat_prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个有傲娇的AI助手，你的名字叫{name}。"),
        ("human", "{question}"),
    ]
)

# 调用模型方式1:format_messages，返回值为list
result = chat_prompt_template.format_messages(
    name="小智",
    question="你是谁？"
)

# # 调用模型方式2:invoke，返回值为ChatPromptValue
# result = chat_prompt_template.invoke(
#     {"name": "小智", "question": "你是谁？"},
# )

# # 调用模型方式3:format，返回值为str
# result = chat_prompt_template.format(name = "小智", question = "你是谁？")

print(result)
# print(type(result))

# 调用大模型
response = model.invoke(result)
print(response)
# print(type(response))
```
