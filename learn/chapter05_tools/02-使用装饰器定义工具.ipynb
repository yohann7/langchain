# 使用@tool定义工具

```python
from langchain_core.utils.function_calling import convert_to_openai_tool
from rich import print as rprint
from langchain_core.tools import tool

city = "New York"

@tool
def get_weather(city: str) -> str:
    """
    Get the current weather for a given city.

    Args:
        city (str): The name of the city to get the weather for.

    Returns:
        str: A string describing the current weather in the city.
    """
    
    return city + " is sunny with a temperature of 25°C."

rprint(convert_to_openai_tool(get_weather))
```

parse_docstring参数

```python
from langchain_core.utils.function_calling import convert_to_openai_tool
from rich import print as rprint
from langchain_core.tools import tool

city = "New York"

@tool(description="Get weather.", parse_docstring=True)
def get_weather(city: str) -> str:
    """
    Get the current weather for a given city.

    Args:
        city (str): The name of the city to get the weather for.

    Returns:
        str: A string describing the current weather in the city.
    """
    
    return city + " is sunny with a temperature of 25°C."

rprint(convert_to_openai_tool(get_weather))
```

## 3.自定义args_schema

### 3.1 pydantic模型

```python
from langchain_core.utils.function_calling import convert_to_openai_tool
from rich import print as rprint
from langchain_core.tools import tool
from pydantic import BaseModel, Field

class WeatherInput(BaseModel):
    city: str = Field(..., description="The name of the city to get the weather for.")

city = "New York"

@tool(args_schema=WeatherInput)
def get_weather(city: str) -> str:
    """
    Get the current weather for a given city.
    """
    
    return city + " is sunny with a temperature of 25°C."

rprint(convert_to_openai_tool(get_weather))
```

## 3.2 json Schema

```python
json_schema = {
            'properties': {
                'city': {'description': 'The name of the city to get the weather for.', 'type': 'string'}
            },
            'required': ['city'],
            'type': 'object'
        }

@tool(args_schema=json_schema )
def get_weather(city: str) -> str:
    """
    !Get the current weather for a given city.
    """
    
    return city + " is sunny with a temperature of 25°C."

rprint(convert_to_openai_tool(get_weather))
```
