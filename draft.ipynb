```python
devices = [
    {"name":"macbook", "color":"silver", "price":1500},
    {"name":"iphone", "color":"blue", "price":1200},
    {"name":"ipad", "color":"gold", "price":1000}
]

print(devices[0]["name"])  # Output: macbook
```

```python
devices = {}
computer = {"name":"macbook", "color":"silver", "price":1500}
phone = {"name":"iphone", "color":"blue", "price":1200}
pad = {"name":"ipad", "color":"gold", "price":1000}

devices["computer"] = computer
devices["phone"] = phone
devices["pad"] = pad
print(devices.get("computer").get("name"))  # Output: macbook
```
